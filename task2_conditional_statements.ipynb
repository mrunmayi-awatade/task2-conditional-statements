```python
# 1. Check if a number is positive
num = 10
if num > 0:
    print("Positive number")

# 2. Check if a number is even or odd
num = 7
if num % 2 == 0:
    print("Even")
else:
    print("Odd")


# 3. Check pass or fail
marks = 65
if marks >= 40:
    print("Pass")
else:
    print("Fail")

# 4. Grade classification
marks = 82
if marks >= 90:
    print("Grade A+")
elif marks >= 75:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")

# 5. Age category
age = 20
if age >= 18:
    print("Adult")
else:
    print("Minor")

# 6. Temperature classification
temperature = 30
if temperature < 15:
    print("Cold")
elif temperature <= 30:
    print("Moderate")
else:
    print("Hot")

# 7. Employee salary category
salary = 50000
if salary >= 50000:
    print("High Salary")
else:
    print("Regular Salary")

# 8. Check voting eligibility
age = 19
if age >= 18:
    print("Eligible")
else:
    print("Not Eligible")

# 9. Student performance
score = 85
if score >= 80:
    print("Excellent")
elif score >= 60:
    print("Good")
else:
    print("Needs Improvement")

# 10. Data value classification
value = 75
if value >= 70:
    print("High")
elif value >= 40:
    print("Medium")
else:
    print("Low")
```

```python
marks = 78
if marks >= 75:
    result = "Distinction"
elif marks >= 60:
    result = "First Class"
elif marks >= 40:
    result = "Pass"
else:
    result = "Fail"
print("Marks:", marks)
print("Classification:", result)
```

```python
age = 20
marks = 85
if age >= 18 and marks >= 40:
    print("Eligible and Passed")
else:
    print("Not Eligible or Failed")
attendance = 65
medical_certificate = True
if attendance >= 75 or medical_certificate:
    print("Allowed to write exam")
else:
    print("Not allowed to write exam")
```
